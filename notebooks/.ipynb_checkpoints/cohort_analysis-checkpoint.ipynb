```python
import sys
import warnings
warnings.filterwarnings("ignore")

sys.path.append("..")

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import seaborn as sns

from utils.helpers import load, explore, parse_dates_inplace

# plot style
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.dpi"] = 130
plt.rcParams["figure.figsize"] = (12,6)

print("✓ imports ok")
```

2- Load & Validate Data

```python
orders = load("orders")
customers = load("customers")
payments = load("payments")
reviews = load("reviews")
```

```python
# Parse all date columns
DATE_COLS = [
    "order_purchase_timestamp"
    "order_approved_at"
    "order_delivered_carrier_date"
    "order_delivered_customer_date"
    "order_estimated_delivery_date"
]
orders = parse_dates_inplace(orders, DATE_COLS)

# Data Quality check
for name, df in [("orders", orders), ("customers", customers), ("payments", payments), ("reviews", reviews)]:
    print(f"{name:12s} → {df.shape[0]:>7,} rows | nulls: {df.isnull().sum().sum()}")
```

```python
# Confirm customer_id is anonymised per order
print("Distinct customer_id    :", customers["customer_id"].nunique())
print("Distinct customer_unique_id:", customers["customer_unique_id"].nunique())
print()
print("→ customer_unique_id is the real repeat-buyer key — always use this!")
```

```python
# Merge orders ← customers to get customer_unique_id
orders = orders.merge(
    customers[["customer_id", "customer_unique_id", "customer_state"]],
    on="customer_id" , how="left"
)

# Merge payments (aggregate to order level)
pay_agg = (payments.groupby("order_id")
           .agg(total_revenue=("payment_value", "sum"),
                n_installments=("payment_installments", "sum")) 
                .reset_index())

orders = orders.merge(pay_agg, on="order_id", how="left")

# keep only delivered orders for cohort analysis
delivered = orders[orders["order_status"].isin(["delivered", "shipped", "invoiced"])].copy()
print(f"Delivered orders: {len(delivered):,}/ {len(orders):,} total")
print(delivered["order_purchase_timestamp"].dtype)  
                                                
```

3- Build Cohort Labels

Each customers is assigned to the month of their first ever purchase. This is called their cohort month (acquisition month).


```python

delivered["order_purchase_timestamp"] = pd.to_datetime(delivered["order_purchase_timestamp"])


# First purchase month per customer
cohort_map = (
    delivered.groupby("customer_unique_id")["order_purchase_timestamp"]
    .min()
    .dt.to_period("M")
    .rename("cohort_month")
    .reset_index()
)

delivered = delivered.merge(cohort_map, on="customer_unique_id", how="left")

# Order month
delivered["order_month"] = delivered["order_purchase_timestamp"].dt.to_period("M")

# Months since first purchase
delivered["month_number"] = (
    delivered["order_month"] - delivered["cohort_month"]
).apply(lambda x: x.n)

print(delivered[["customer_unique_id", "cohort_month", "order_month", "month_number"]].head(8))
```

```python
# Cohort sizes (how many customers acquired each month )
cohort_sizes = (
    delivered[delivered["month_number"] == 0]
    .groupby("cohort_month")["customer_unique_id"]
    .nunique()
    .rename("cohort_size")
)

print(f"Total cohorts : {len(cohort_sizes)}")
print(f"Date range  : {cohort_sizes.index.min()}) → {cohort_sizes.index.max()}")
print()
print(cohort_sizes.sort_index().head(10))
```

4- Retention Table

Count how many unique customers from each cohort returned in each subsequent month.
Then divide by cohort size to get the retention rate %

```python
# Raw retention counts
retention_counts = (
    delivered.groupby(["cohort_month", "month_number"])["customer_unique_id"]
    .nunique()
    .reset_index()
    .rename(columns={"customer_unique_id": "retained_customers"})
)

# Pivot: rows = cohort_month, cols = month_number
retention_pivot = retention_counts.pivot_table(
    index="cohort_month",
    columns="month_number",
    values="retained_customers"
)

# Retention rate % = retained / cohort_size (Month 0)
retention_rate = retention_pivot.divide(cohort_sizes, axis=0) * 100
retention_rate = retention_rate.round(2)

print("Retention table (%) - first 5 cohorts, first 6 months:")
print(retention_rate.iloc[:5, :6])
```

5- Cohort Retention Heatmap

- Rows = acquisition (cohort)
- Columns = months since first purchase (0 = acquisition month)
- Color = retention rate % (darker = higher retention)

```python
fig, ax = plt.subplots(figsize=(16, 9))

# Mask NaN cells
mask= retention_rate.isnull()

sns.heatmap(
    retention_rate,
    mask=mask,
    annot=True,
    fmt=".1f",
    linewidths=0.4,
    linecolor="white",
    cmap="YlOrRd",
    vmin=0, vmax=100,
    cbar_kws={"label": "Retention Rate (%)"},
    ax=ax
)

ax.set_title("Customer Cohort Retention Heatmap\nOlist E-Commerce 2016-2018",
             fontsize=15, fontweight="bold", pad=14)
ax.set_xlabel("Months Since first Purchase", fontsize=11)
ax.set_ylabel("Cohort (Acquisition Month)", fontsize=11)
ax.set_yticklabels([str(y.get_text()) for y in ax.get_yticklabels()], fontsize=8, rotation=0)

plt.tight_layout()
plt.savefig("../reports/cohort_heatmap.png", dpi=150, bbox_inches="tight")
plt.show()
print("✓ saved → reports/cohort_heatmap.png")
```

```python
# CEO answer: 2017 Cohohrts retention in 2018
cohorts_2017 = retention_rate[retention_rate.index.year == 2017]
print("=== 2017 Cohorts - Retention into 2018 ===")
print(cohorts_2017.iloc[:, :13]) #Month 0 to Month 12 
print()
avg_m12 = cohorts_2017[12].mean() if 12 in cohorts_2017.columns else None
if avg_m12:
    print(f"Average retention at Month 12 (2017→2018): {avg_m12:.2f}%")
```

6- Bonus:Deliverly Time vs Review score per state

Do states with slower deliverly get worse reviews?

```python
# Convert both date columns 
orders["order_delivered_customer_date"] = pd.to_datetime(orders["order_delivered_customer_date"])
orders["order_purchase_timestamp"] = pd.to_datetime(orders["order_purchase_timestamp"])


# Delivery days
orders["delivery_days"] = (
    orders["order_delivered_customer_date"] - orders["order_purchase_timestamp"]
).dt.days

# Merge reviews
rev = reviews[["order_id", "review_score"]]
analysis = orders.merge(rev, on="order_id", how="left")
analysis = analysis[
    (analysis["order_status"] == "delivered") &
    (analysis["delivery_days"] > 0) &
    (analysis["delivery_days"].notna())
]

state_stats = (
    analysis.groupby("customer_state")
    .agg(avg_delivery_days=("delivery_days", "mean"),
        avg_review_score=("review_score", "mean"),
        n_orders=("order_id", "count"))
    .round(2)
    .sort_values("avg_delivery_days",ascending=False)
    .reset_index()
)

print(state_stats.head(10))
```

```python
fig, axes = plt.subplots(1,2, figsize=(16,6))

# Left: delivery dayps per state (top 15 slowest)
top15 = state_stats.head(15)
axes[0].barh(top15["customer_state"], top15["avg_delivery_days"], color="#e07b54")
axes[0].set_xlabel("Avg Delivery Days")
axes[0].set_title("Top 15 Slowest states")
axes[0].invert_yaxis()

# Right: scatter - delivery days vs review score 
axes[1].scatter(state_stats["avg_delivery_days"], state_stats["avg_review_score"],
                s=state_stats["n_orders"] / 30, alpha=0.7, color="#4c72b0", edgecolors="white")
for _, row in state_stats.iterrows():
    axes[1].annotate(row["customer_state"],
                     (row["avg_delivery_days"], row["avg_review_score"]),
                     fontsize=6, alpha=0.7)
axes[1].set_xlabel("Avg Delivery Days")
axes[1].set_ylabel("Avg Review Score")
axes[1].set_title("Delivery Time vs Review Score per State")

# Correlation
corr = state_stats["avg_delivery_days"].corr(state_stats["avg_review_score"])
axes[1].set_title(f"Delivery Time vs Review Score (r = {corr:.2f})")

plt.tight_layout()
plt.savefig("../reports/delivery_vs_review.png", dpi=150, bbox_inches="tight")
plt.show()
print(f"✓ Correlation: {corr:.2f} — {'negative as expected' if corr < 0 else 'check data'}")

```

7- Bonus: Power Users (Top 1% by Revenue)

Identify the highest-value customers by lifetime spend.


```python
clv = (
    delivered.groupby("customer_unique_id")
    .agg(lifetime_revenue=("total_revenue", "sum"),
         total_orders=("order_id", "count"),
         first_order=("order_purchase_timestamp", "min"),
         last_order=("order_purchase_timestamp", "max"))
    .reset_index()
    .sort_values("lifetime_revenue", ascending=False)
)

p99 = clv["lifetime_revenue"].quantile(0.99)
power_users = clv[clv["lifetime_revenue"] >= p99].copy()

print(f"P99 threshold  : R$ {p99:,.2f}")
print(f"Power users    : {len(power_users):,} customers")
print(f"% of base      : {len(power_users)/len(clv)*100:.2f}%")
print(f"Their revenue  : R$ {power_users['lifetime_revenue'].sum():,.2f}")
print(f"Total revenue  : R$ {clv['lifetime_revenue'].sum():,.2f}")
print(f"Revenue share  : {power_users['lifetime_revenue'].sum()/clv['lifetime_revenue'].sum()*100:.1f}%")
print()
print(power_users.head(10).to_string(index=False))


```
